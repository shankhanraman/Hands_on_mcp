# Tools
Models can request to call tools that perform tasks such as fetching from a database , searching the web or running code. Tools are pairing of :
1. A schema including the name of tool , a descirption and /or argument definintions (often a JSON schema) 
2. A function or co routing to execute 
 

```python
import os
from dotenv import load_dotenv
load_dotenv
os.environ["OPENAI_API_KEY"]= os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]= os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]= os.getenv("GROQ_API_KEY")
```

```python
from langchain_openai import ChatOpenAI
model=ChatOpenAI(model_name="gpt-4.1", temperature=0.5)
response = model.invoke("Hello How are you")
```

```python
from langchain.tools import tool
@tool
def get_weather(location:str)-> str:
    """ Get the current weather at a location """
    return f"It is sunny in {location}  "

model_with_tools = model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("What is the weather in New York?")
print(response)

for tool_call in response.tool_calls:
    print(f"Tool called: {tool_call['name']} ")
    print(f"Tool Args: {tool_call['args']}")
```

### Tools Exectuon Loops

```python
# Step 1 :  Model genrates tool calls
messages = [{"role":"user","content":"What is the weather in Boston"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)


# Step 2 : Exectue the tools and collect results 
for tool_call in ai_msg.tool_calls:
    # Execute the tool with generated arguments
    tool_result = get_weather.invoke(tool_call)
    messages.append(tool_result)

# Step 3 :  Pass results back to model for the final response
final_response = model_with_tools.invoke(messages)
print(final_response.text)

# THe current weather in Boston is 73 F and sunny
```
