# Streaming
Most models can stream their output content while it is being generated. BY displaying output progressively , streaming signifcantly imporves user experinec , partilcuary for longer responses . Claiing stream() resturn as iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real time 

```python
import os
from langchain.chat_models  import init_chat_model
os.environ["GOOGLE_API_KEY"]= os.getenv("GOOGLE_API_KEY")

model = init_chat_model("google_genai:gemini-3.5-flash")
response = model.invoke("Why do parrots talk?")
response.content
```

```python
model.stream("Write me a 200 words in Artifical intelligence")
```

```python
for chunk in model.stream("Write me a 200 words in Artifical intelligence"):
    print(chunk.text)
```

```python
for chunk in model.stream("Write me a 200 words in Artifical intelligence"):
    print(chunk.text,end="|",flush=True)
```

# Batch
Batching is collection of independent requests to a model can significantly improve performance and reduce costs , as the processing can be done in parallel

```python
responses = model.batch([
    "Why do parrots talk?",
    "Why do airplanes fly",
    "What is uique about quantum computing"

])
for response in responses:
    print(response)
```

```python
responses = model.batch([
    "Why do parrots talk?",
    "Why do airplanes fly",
    "What is uique about quantum computing"]
    config={
        'max_concurrency':5 # Limit to 5 parallel calls
    }
    )
for response in responses:
    print(response)
```
