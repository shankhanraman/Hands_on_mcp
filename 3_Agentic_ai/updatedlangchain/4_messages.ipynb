# Messages 
Messages are the fundamental unit of context for models in Langchain. They represent the input and output of models , carrying both the content and metadata needed to represent the state of conversion when interacting with an LLM.Messages are objects that contain:
. Role - Identifies the message type (eg system user)
. Content - Represent the actual content of the message (like text messages images audio documents)
. Metadata- Optional fields such as response information , message IDs and token usage
Langchain provides a standard  message type that works across all mdoel providers , esuring cosotent behaviour regradless of the model nbeing called 

```python
import os
from langchain.chat_models import init_chat_model

os.environ["OPENAI_API_KEY"]= os.getenv("OPENAI_API_KEY")

model = init_chat_model("gpt-4.1")
```

## Text prompts

Use text prompt when:

## Message prompts
. System message - Tells the model how to behave and provide context for interactions
. Human Message - represents user input and interations with model 
. AI message _ reposnes genreated by model including text content tools calls and metadata
. Tools message - Represnet the output of tool calls

## System Message 
A Sytem message represant a intial set of instructions that primes the model's behaviour . You can use a sytem to set the tone and define the model role and establish guideline for responses

## Human Message 
A Human message represnt user input and interactions. They can contain text, images , aduio , files and any other amount of multi modal content.

## AI Message 
An AI message represent the output of a model invocagtion. They can include multimodal data , tool calls qand provider-specific metadata that you can later access

## Tool Message 
For models that ssupport tool calling AI messages can contain tool calls . Tool messages are used to pass the results of single tool execution back to a model


```python
from langchain.messages import SystemMessage,HumanMessage,AIMessage
messages=[
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write a poem on Artifical Intelligencde")  
]
response = model.invoke(messages)
response.content
```

```python
system_message = SystemMessage("You are a helpful coding Assistant")

messages=[
    system_message,
    HumanMessage("How do I create a REST API")
]
response = model.invoke(messages)
print(response.content)
```

```python
from langchain.messages import AIMessage,SystemMessage,HumanMessage

#Create an AI message manually (e.g for conversation History)
ai_msg = AIMessage("I d be happy to help you with that question")

# Create an AI message Manually (e.g for conversation HIstory)
messages =[
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me"),
    ai_msg, # Insert as if it came from model
    HumanMessage("Great what is 2+2")
]

response = model.invoke(messages)
print(response.content)
```

```python
from langchain.messages import AIMessage
from langchain.messages import ToolMessage

# After a model makes a tool call
# (Here, we demonstrate manually creating the messages for brevity)
ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_123"
    }]
)

# Execute tool and create result message
weather_result = "Sunny, 72°F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"  # Must match the call ID
)

# Continue conversation
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_message,  # Model's tool call
    tool_message,  # Tool execution result
]
response = model.invoke(messages)  # Model processes the result
```
